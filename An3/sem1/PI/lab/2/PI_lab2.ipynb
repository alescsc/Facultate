```python
import torch

x = torch.tensor([-10.0, 10.0, 8.0], requires_grad=True)
y = 1/(1 + torch.exp(-x))
y.sum().backward()

print(x.grad)
```

```python
import torch
A = torch.tensor([[1, 2, 3], [4, 5, 6]]).float()
B = torch.tensor([[10, 11], [-2, 0], [1, -3]]).float()

print(torch.mm(A, B))
print(torch.norm(A))
print(torch.norm(B))
```
